Подключение библиотек и загрузка датасета:

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sb
```

```python
df = pd.read_csv('imdb_top_1000.csv')
df.head(10)
```

Буду работать с рейтингом фильмов:

```python
rating_column = 'IMDB_Rating'
```

Построение и загрузка графика в виде файла:

```python
plt.figure(figsize=(10, 6))
plt.hist(df[rating_column], bins=20, edgecolor='black', alpha=0.7, color='steelblue')
plt.title('IMDB Rating distribution', fontsize=14)
plt.xlabel('Rating', fontsize=12)
plt.ylabel('Film number', fontsize=12)
plt.grid(axis='y', alpha=0.3)
plt.savefig('filmRating_distribution.png')
plt.show()
```
